```python
import matplotlib.pyplot as plt
import numpy as np

delays = np.linspace(0, 365, 100)
```

```python
def subjective_value(amount, delay, k):
    value = amount / (1 + k * delay)
    return value

k = 0.01
values = []

for delay in delays:
    value = subjective_value(100, delay, k)
    values.append(value)
```

```python
plt.plot(delays, values)

plt.xlabel("Delay (days)")
plt.ylabel("Subjective Value ($)")
plt.title("Hyperbolic Delay Discounting")

plt.show()
```

For a fixed $100 reward and discount rate k = 0.01, subjective value decreases as delay increases. The decline is steepest at shorter delays and gradually flattens, which is characteristic of hyperbolic delay discounting.

```python
def subjective_value(amount, delay, k):
    value = amount / (1 + k * delay)
    return value

delays = [0, 7, 30, 90, 180, 365]
k_values = [0.001, 0.01, 0.05, 0.1]

for k_value in k_values:
    values = []

    for delay in delays:
        value = subjective_value(100, delay, k_value)
        values.append(value)

    plt.plot(delays, values, label=f"k = {k_value}")

plt.xlabel("Delay (days)")
plt.ylabel("Subjective Value ($)")
plt.title("Hyperbolic Delay Discounting")
plt.legend()
plt.show()
```

Interpretation:

The hyperbolic discounting model predicts that the subjective value of a delayed reward decreases as the delay increases. Larger values of k produce steeper discounting curves, meaning that the same future reward is valued much less in the present. Individuals with larger estimated k values would therefore be expected to show a stronger preference for smaller, sooner rewards.
