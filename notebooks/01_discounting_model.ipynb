# Hyperbolic Delay Discounting

## Question

How does the subjective value of a reward change as the delay to receiving it
increases?

This notebook introduces the hyperbolic delay-discounting model used throughout
the project:

\[
V = \frac{A}{1 + kD}
\]

where:

- `V` is subjective value
- `A` is reward amount
- `D` is delay
- `k` is the discount-rate parameter

Larger values of `k` indicate steeper discounting of delayed rewards.

```python
import matplotlib.pyplot as plt
import numpy as np

delays = np.linspace(0, 365, 100)
```

```python
def subjective_value(amount, delay, k):
    value = amount / (1 + k * delay)
    return value

k = 0.01
values = []

for delay in delays:
    value = subjective_value(100, delay, k)
    values.append(value)
```

## Discounting a Single Reward

First, consider a $100 reward discounted across delays from 0 to 365 days
using `k = 0.01`.

```python
plt.plot(delays, values)

plt.xlabel("Delay (days)")
plt.ylabel("Subjective Value ($)")
plt.title("Hyperbolic Delay Discounting")

plt.show()
```

For a fixed $100 reward and discount rate k = 0.01, subjective value decreases as delay increases. The decline is steepest at shorter delays and gradually flattens, which is characteristic of hyperbolic delay discounting.

## Effect of the Discount Rate

The discount-rate parameter `k` controls how quickly delayed rewards lose
subjective value. Comparing several values of `k` illustrates how individual
differences in discounting affect the shape of the value function.

```python
k_values = [0.001, 0.01, 0.05, 0.1]

for k_value in k_values:
    values = []

    for delay in delays:
        value = subjective_value(100, delay, k_value)
        values.append(value)

    plt.plot(delays, values, label=f"k = {k_value}")

plt.xlabel("Delay (days)")
plt.ylabel("Subjective Value ($)")
plt.title("Effect of Discount Rate on Subjective Value")
plt.legend()
plt.show()
```

## Interpretation

The hyperbolic discounting model predicts that the subjective value of a delayed reward decreases as the delay increases. Larger values of k produce steeper discounting curves, meaning that the same future reward is valued much less in the present. Individuals with larger estimated k values would therefore be expected to show a stronger preference for smaller, sooner rewards.

## Conclusion

Hyperbolic delay discounting describes how the subjective value of a reward
decreases as the delay to receiving it increases.

The parameter `k` controls the steepness of this decline. Larger values of
`k` produce more rapid discounting, meaning delayed rewards lose subjective
value more quickly.
